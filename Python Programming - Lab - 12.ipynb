### Practical 1: Count Lines, Words and Characters

**Problem Definition:**  
Write a Python program to read text from a file and count the total number of lines, words, and characters. The result should be written to an output file.


```python
f1 = open("file1.txt","r")
f2 = open("file2.txt","a")

l = f1.readlines()
f2.write("line:"+str(len(l))+"\n")
f1.seek(0)
f2.write("word:"+str(len(f1.read().split()))+"\n")
f1.seek(0)
f2.write("charcter:"+str(len(f1.read()))+"\n")

f1.close()
f2.close()
```


### Practical 2: Search a Word in a File

**Problem Definition:**  
Write a Python program to search a given word in a text file and count its occurrences. Store the result in an output file.


```python
f1 = open("file1.txt","r")
f2 = open("file3.txt","w")
w1 = input("Enter a Word:")
w2 = f1.read().split()
count = 0
for i in w2:
    print(i)
    if(i == w1):
        count+=1

f2.write(w1+" "+str(count))
f1.close()
f2.close()
```


### Practical 3: Student Marks and Average

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Store marks in a list, calculate the average using a user-defined function, and write student details along with average marks to an output file.


```python
def avg(marks):   
    return sum(marks)/len(marks)

f1 = open("file4.txt","r")
f2 = open("file5.txt","w")
for i in f1:
    data = i.split()
    marks = data[2:]
    marks = list(map(int , marks))
    print(marks)
    # av = avg(marks)
    f2.write(str(avg(marks)) end= " ")
    
f1.close()
f2.close()


```


### Practical 4: Maximum and Minimum using Tuple

**Problem Definition:**  
Write a Python program to read numbers from a file, store them in a tuple, and find the maximum and minimum values. Write the result to an output file.


```python
f1 = open("file6.txt","r")
f2 = open("file7.txt","w")
number = f1.read().split(",")
print(tuple(number))
min_num = int(number[0])
max_num = int(number[0])
for i in number:
    if int(i) < min_num:
        min_num = int(i)
    elif int(i) > max_num:
        max_num = int(i)
f2.write("min:"+str(min_num) + " " + "max:"+str(max_num))
f1.close()
f2.close()
```


### Practical 6: Word Frequency using Dictionary

**Problem Definition:**  
Write a Python program to count the frequency of each word in a text file using a dictionary and write the result to an output file.


```python
f1 = open("file1.txt","r")
text = f1.read()
f1.close()
words = text.lower().split()
freq = {}
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1
f2 = open("file2.txt","a")
for word in freq:
    f2.write("Word :"+str(freq[word]) + "\n")
f2.close()
print("Done!")
```


### Practical 7: Student Result and Grade Calculation

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Use a user-defined function to determine grade (A/B/C/Fail) and write the student result to an output file.

**Grade Criteria:**  
- Marks ≥ 75 : Grade A  
- Marks ≥ 60 : Grade B  
- Marks ≥ 40 : Grade C  
- Marks < 40 : Fail


```python
def grade(marks):
    if marks >= 75:
        return 'A'
    elif marks >= 60:
        return 'B'
    elif marks >= 40:
        return 'C'
    else:
        return 'Fail'
f1 = open("file4.txt","r")
f2 = open("file5.txt","w")
for line in f1:
    data = line.split()
    roll = data[0]
    name = data[1]
    m1 = int(data[2])
    m2 = int(data[3])
    m3 = int(data[4])
    total = m1+m2+m3
    avg = total/3
    g = grade(avg)
    f2.write(f"{roll} {name} Total:{total} Avg:{avg:.2f} Grade:{g}\n")
f1.close()
f2.close()
```
